# Find and Bind Pair

Demonstration: Two agents find each other as loved ones and bind.

```python
from gra_love_null.agents import LoveNullAgent
from gra_love_null.semantics import InternalEmbeddingProvider, semantic_similarity
import numpy as np

# Create two agents with similar memory vectors
base_vector = np.random.randn(128) * 0.1

agent_a = LoveNullAgent(id="alice", role="explorer", task="map_territory")
agent_a.memory_vector = base_vector.copy()

agent_b = LoveNullAgent(id="bob", role="explorer", task="map_territory")
agent_b.memory_vector = base_vector + np.random.randn(128) * 0.05  # Similar but not identical

# Create provider
provider = InternalEmbeddingProvider(dim=128)

# Compute similarity
sim = semantic_similarity(agent_a, agent_b, provider)
print(f"Semantic similarity between Alice and Bob: {sim:.4f}")

# Discover love objects
population = [agent_a, agent_b]
loved_a = agent_a.discover_love_objects(population, provider, strategy="top_k_nearest", k=1)
loved_b = agent_b.discover_love_objects(population, provider, strategy="top_k_nearest", k=1)

print(f"Alice loves: {[a.id for a, _ in loved_a]}")
print(f"Bob loves: {[a.id for a, _ in loved_b]}")

# Check mutual love
is_mutual = (
    len(loved_a) > 0 and loved_a[0][0].id == "bob" and
    len(loved_b) > 0 and loved_b[0][0].id == "alice"
)
print(f"Mutual love: {is_mutual}")
```

```python
# Simulate life with love
for step in range(10):
    result_a = agent_a.step()
    result_b = agent_b.step()
    
    print(f"Step {step}: "
          f"Alice energy={agent_a.affect.energy:.2f}, burnout={agent_a.affect.burnout:.2f}, "
          f"love={agent_a.affect.love_strength:.2f}")
    print(f"          "
          f"Bob energy={agent_b.affect.energy:.2f}, burnout={agent_b.affect.burnout:.2f}, "
          f"love={agent_b.affect.love_strength:.2f}")
```
