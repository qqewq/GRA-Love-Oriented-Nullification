# Ossetian Cluster Love

Demonstration: Agents form semantic clusters and love their cluster centroids ("elder siblings").

```python
from gra_love_null.agents import LoveNullAgent
from gra_love_null.semantics import InternalEmbeddingProvider, cluster_agents, find_cluster_centroids
from gra_love_null.love_selection import select_by_cluster_centroid
import numpy as np

# Create diverse agent population
n_agents = 30
agents = []
for i in range(n_agents):
    agent = LoveNullAgent(id=f"a{i}", role="worker", task=f"task_{i % 5}")
    # Create clusters by assigning similar vectors
    cluster_id = i // 10  # 3 clusters
    agent.memory_vector = np.random.randn(128) * 0.1
    agent.memory_vector[cluster_id * 40:(cluster_id + 1) * 40] += 1.0
    agents.append(agent)

provider = InternalEmbeddingProvider(dim=128)

# Cluster agents
clusters = cluster_agents(agents, provider, n_clusters=3)
centroids = find_cluster_centroids(agents, clusters, provider)

print("Clusters:")
for cid, indices in clusters.items():
    centroid_idx = centroids[cid]
    print(f"  Cluster {cid}: {len(indices)} agents, centroid = {agents[centroid_idx].id}")

# Each non-centroid agent loves its cluster centroid
for cluster_id, centroid_idx in centroids.items():
    centroid = agents[centroid_idx]
    for idx in clusters[cluster_id]:
        if idx != centroid_idx:
            agent = agents[idx]
            # Love the centroid as "elder sibling"
            agent.loved_ids.append(centroid.id)
            agent.loved_agents.append(centroid)
            print(f"  {agent.id} loves centroid {centroid.id}")
```
